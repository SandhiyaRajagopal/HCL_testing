Print all prime numbers between input range (Ex – input 20 50, prints all prime numbers between 20 and 50).

```python
start=int(input())
end=int(input())
for n in range(start,end+1):
    count=0
    for i in range(1,n+1):
        if(n % i == 0):
            count+=1
    if count==2:
        print(n,end=" ")
```

Factorial using recursion

```python
n=int(input())
def fact(n):
    if n==0 or n==1:
        return 1
    else:
        return n * fact(n-1)
print(fact(n))
```

```python
s=input()
freq={}
for ch in s:
    if ch in freq:
        freq[ch]=freq[ch]+1
    else:
        freq[ch]=1
for ch in freq:
    print(ch,freq[ch])
        
```

    if n==0:
        return 0
    elif n==1:
        return 1

# PYTHON FUNCTION ASSIGNMENT

1.Write a function calculate(a, b, operation) that performs addition, subtraction, multiplication, or division based on the supplied operation.

```python
def calculate(a, b, operation):
    if operation == "+":
        return a + b
    elif operation == "-":
        return a - b
    elif operation == "*":
        return a * b
    elif operation == "/":
        return a / b

a = int(input())
b = int(input())
operation = input()
print(calculate(a, b, operation))
```

2.Write a function sum_numbers(*args) that accepts any number of arguments and returns their sum.

```python
def sum_numbers(*args):
    total = 0
    for x in args:
        total = total + x
    return total
n = int(input())
a = list(map(int, input().split()))
print(sum_numbers(*a))
```

3.Write a function employee(**args) that accepts employee information such as name, ID, department and salary, then displays the information.

```python
def employee(**args):
    for key in args:
        print(key, ":", args[key])
name = input()
id = input()
department = input()
salary = input()
employee(name=name, ID=id, department=department, salary=salary)
```

4.Write a function remove_duplicates(lst) that returns a list containing only unique elements while preserving their original order.

```python
def remove_duplicates(lst):
    result = []
    for x in lst:
        if x not in result:
            result.append(x)
    return result
n = int(input())
lst = list(map(int, input().split()))
print(remove_duplicates(lst))
```

5.Using a lambda function, sort a list of tuples based on the second element. Example: [(1,5), (2,3), (4,1)].

```python
n = int(input())
lst = []
for i in range(n):
    a, b = map(int, input().split())
    lst.append((a, b))
lst.sort(key=lambda x: x[1])
print(lst)
```

Student Performance Analysis

```python
marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])
total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
highest = np.max(marks, axis=1)
lowest = np.min(marks, axis=1)
class_average = np.mean(marks)
print("Marks:")
print(marks)
print("\nStudent Total:", total)
print("Student Average:", average)
print("Highest mark:", highest)
print("Lowest mark:", lowest)
print("\nClass Average:", class_average)
print("Students above class average:")

for i in range(len(average)):
    if average[i] > class_average:
        print("Student", i + 1)
```
